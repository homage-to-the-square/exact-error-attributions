# Aggregated Density Plots (2 x 3): Realized Efficiency & Data Defect Correlation by Sampling Scheme

```python
schemes = [
    ("Biased in Y", df_y_bias),
    ("Biased in X", df_x_bias),
    ("Biased in Abs Y", df_abs_y_bias),
]
label_map = {
    'srs_corr': 'Unbiased', 'biased_corr': 'Biased',
    'srs_fi': 'Unbiased', 'biased_fi': 'Biased',
}

fig, axes = plt.subplots(2, 3, constrained_layout=True, figsize=(18, 8), sharey='row')
legend_handles, legend_labels = None, None

for col, (title, df) in enumerate(schemes):
    ax = axes[0, col]
    sns.kdeplot(df[['srs_corr', 'biased_corr']], fill=True, ax=ax)
    handles, labels = ax.get_legend_handles_labels()
    if legend_handles is None:
        legend_handles, legend_labels = handles, [label_map[l] for l in labels]
    ax.get_legend().remove()
    ax.set_title(title)
    ax.set_xlabel('Data Defect Correlation')
    ax.set_ylabel('Density' if col == 0 else '')

    ax = axes[1, col]
    sns.kdeplot(df[['srs_fi', 'biased_fi']], fill=True, ax=ax)
    ax.get_legend().remove()
    ax.set_xlabel(Jn_name)
    ax.set_ylabel('Density' if col == 0 else '')

fig.legend(legend_handles, legend_labels, loc="upper center", bbox_to_anchor=(0.5, 1.06), ncol=2)
plt.savefig(f'aggregated_densities_2x3{savefile_suffix}.png', bbox_inches='tight')
```

# Aggregated Joint Density Plots (1 x 4): Ideal Sample vs Biased Samples

```python
fig, axes = plt.subplots(1, 4, constrained_layout=True, figsize=(20, 5))

ax = axes[0]
sns.kdeplot(
    data=df_y_bias,
    x="srs_fi",
    y="srs_corr",
    fill=True,
    bw_adjust=1.8,
    levels=5,
    ax=ax,
)
ax.set_xlabel("Ideal Sample")
ax.set_ylabel("")

biased_schemes = [
    ("Biased in Y", df_y_bias),
    ("Biased in X", df_x_bias),
    ("Biased in Abs Y", df_abs_y_bias),
]

for i, (title, df) in enumerate(biased_schemes, start=1):
    ax = axes[i]
    sns.kdeplot(
        data=df,
        x="biased_fi",
        y="biased_corr",
        fill=True,
        bw_adjust=1.8,
        levels=5,
        ax=ax,
    )
    ax.set_xlabel(title)
    ax.set_ylabel("")

fig.supylabel("Data Defect Correlation")
fig.supxlabel(f"{Jn_name} (Across Sampling Mechanisms)")
plt.savefig(f"joint_density_1x4{savefile_suffix}.png", bbox_inches="tight")
```
